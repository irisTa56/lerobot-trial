https://rerun.io/docs/reference/dataframes

```python
from pathlib import Path

import rerun as rr
from ipyfilechooser import FileChooser
```

```python
file_chooser = FileChooser(
    placeholder="Select .rrd file",
    description="RRD File:",
    filter_pattern="*.rrd",
)

file_chooser
```

```python
rrd_path = Path(file_chooser.selected) if file_chooser.selected else None
print(f"Selected: {rrd_path}")
```

```python
recording = rr.dataframe.load_recording(rrd_path)
schema = recording.schema()

display(schema.index_columns())
display(schema.component_columns())
```

```python
view = recording.view(index="log_time", contents="/**")
df = view.select().read_pandas()
df
```
