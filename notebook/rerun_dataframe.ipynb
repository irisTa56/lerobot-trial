```python
from pathlib import Path

import rerun as rr
from ipyfilechooser import FileChooser
```

```python
file_chooser = FileChooser(
    placeholder="Select .rrd file",
    description="RRD File:",
    filter_pattern="*.rrd",
)

file_chooser
```

```python
rrd_path = Path(file_chooser.selected) if file_chooser.selected else None
print(f"Selected: {rrd_path}")
```

```python
with rr.server.Server(datasets={"my_dataset": [rrd_path]}) as server:
    client = server.client()
    dataset = client.get_dataset("my_dataset")

    schema = dataset.schema()
    display(schema.index_columns())
    display(schema.component_columns())

    view = dataset.filter_contents("/**")
    df = view.reader(index="log_time").to_pandas()
```
